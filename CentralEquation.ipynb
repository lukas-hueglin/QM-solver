```python
import matplotlib as mpl
import matplotlib.pyplot as plt

import pickle as pkl

from qmsolver.CentralEquation import *

mpl.rcParams['text.usetex'] = True
```

```python
ce = pkl.load('CentralEquation.pkl')
```

```python
ce = CentralEquation(
    basis_vectors=[[1, 1, 0], [0, 1, 1], [1, 0, 1]],
    atoms = [[0, 0, 0]],
    atom_potential = lambda r: -e / (4*np.pi*eps_0*np.linalg.norm(r, axis=-1)),
    num_cells=51,
    res_cells=11
)
```

```python
verts, edges, faces = ce.wignerseitz_cell()
ce.register_symmetries(
    {
        'G': (r'\(\Gamma\)', [0, 0, 0]),
        'W': (r'\(W\)', verts[0][0]),
        
        'K': (r'\(K\)', edges[0][0]),
        'U': (r'\(U\)', edges[0][1]),

        'X': (r'\(X\)', faces[2]),
        'L': (r'\(L\)', faces[1])
    }
)
```

```python
path, ticks, tick_labels = ce.build_symmetry_path(['L', 'G', 'X'], 30)
ce.solve_central_equation(path)
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2, projection='3d')

ax1.plot(ce.E[:, :10])
ax2.plot(*get_coords(path), color='red')
ce.plot_wignerseitz_cell(ax2)

ax1.set_xticks(ticks)
ax1.set_xticklabels(tick_labels)

ax2.set_xlim((-5, 5))
ax2.set_ylim((-5, 5))
ax2.set_zlim((-5, 5))
ax2.set_aspect('equal')
plt.show()
```
